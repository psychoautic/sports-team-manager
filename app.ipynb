### We Start By creating Classes. For our sport team Manager we decided to have the following classes:
- Team
- Player
- Coach
- Manager
- Match
- Tournament 

```python
from typing import List, Optional, Dict

class Team:
    def __init__(self, name: str, manager: str, coach: str, players: List['Player']):
        self.name: str = name
        self.manager: str = manager
        self.coach: str = coach
        self.players: List['Player'] = players

class Player:
    def __init__(self, name: str, position: str, number: int, team: Optional['Team'] = None, stats: Optional[Dict[str, int]] = None):
        self.name: str = name
        self.position: str = position
        self.number: int = number
        self.team: Optional['Team'] = team
        self.stats: Optional[Dict[str, int]] = stats

class Coach:
    def __init__(self, name: str, team: Optional['Team'] = None):
        self.name: str = name
        self.team: Optional['Team'] = team


class Manager:
    def __init__(self, name: str, team: Optional['Team'] = None):
        self.name: str = name
        self.team: Optional['Team'] = team

class Match:
    def __init__(self, team_a: 'Team', team_b: 'Team', date: str = None, location: str = None, result: dict = None):
        self.team_a: 'Team' = team_a
        self.team_b: 'Team' = team_b
        self.date: str = date
        self.location: str = location
        self.result: dict = result if result else {'team_a': 0, 'team_b': 0}


    def update_result(self, team_a_score: int, team_b_score: int):
        self.result['team_a'] = team_a_score
        self.result['team_b'] = team_b_score

class League:
    def __init__(self, name: str, teams: List['Team'] = None):
        self.name: str = name
        self.teams: List['Team'] = teams if teams else []
```
